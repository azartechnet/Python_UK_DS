```python
# Stacking is same as concatenation the only different is that stacking is done along a new axis
```

```python
import numpy as np
```

```python
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
arr=np.stack((arr1,arr2),axis=1)
print(arr)
```

```python
#Stack along with rows
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
arr=np.hstack((arr1,arr2))
print(arr)
```

```python
#Stack along with columns
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
arr=np.vstack((arr1,arr2))
print(arr)
```

```python
#Stacking along height(Depth)
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
arr=np.dstack((arr1,arr2))
print(arr)
```

```python
#Sentence Tokenizations
from nltk import word_tokenize,sent_tokenize
sent="Welcome to nlp application"
print(word_tokenize(sent))
print(sent_tokenize(sent))
```

```python
#Stemming
from nltk.stem import PorterStemmer

porter=PorterStemmer()
print(porter.stem("play"))
print(porter.stem("playing"))
print(porter.stem("played"))
print(porter.stem("plays"))
print(porter.stem("Communication"))
print(porter.stem('Books'))
print(porter.stem('Booking'))
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer=WordNetLemmatizer()
print(lemmatizer.lemmatize("Communication"))
```

```python
#Removing stop words with NLTK
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
example_sent="This was as a Sample hat"
stop_words=set(stopwords.words('english'))
word_tokens=word_tokenize(example_sent)
#print(word_tokens)
#Convert the words in word_tokens to lowercase and then checks whether
#they are present in the stop_words or not
filtered_sentence=[w for w in word_tokens if not w.lower() in stop_words]
filtered_sentence=[]

for w in word_tokens:
    if w not in stop_words:
        filtered_sentence.append(w)

print(word_tokens)
print(filtered_sentence)#Stop word is an Upper case it will show
```

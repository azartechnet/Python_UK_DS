```python
#Pandas Series
#What is a Series?
#A pandas series is like a column in a table
#It is one dimensional array holding data of any type
```

```python
import pandas as pd
a=[1,7,10]
my1=pd.Series(a)
print(my1)
```

```python
print(my1[0])
```

```python
my1=pd.Series(a,index=["x","y","z"])
print(my1)
```

```python
print(my1["x"])
```

```python
my2={101:"azar",102:"mohamed",103:"sai"}
result=pd.Series(my2,index=[101,102])
print(result)
```

```python
#DataFrames
#Data sets in pandas are usually multi-dimensional tables called dataFrames
#Series is like a column,a dataframe is the whole table
```

```python
data={"c1":[100,200,300],"c2":[400,500,600]}
my1=pd.DataFrame(data)
print(my1)
```

```python
#Locate Row
#Pandas use the loc attribute to return one or more specified rows
```

```python
print(my1.loc[0])
```

```python
#Use a list of indexes
print(my1.loc[[0,1]])
```

```python
#Named indexes
#With the index argument you can name your own indexes
```

```python
my1
```

```python
my2=pd.DataFrame(data,index=["d1","d2","d3"])
print(my2)
```

```python
#Locate Named Indexes
#Refer to the named index
print(my2.loc["d1"])
```

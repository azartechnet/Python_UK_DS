```python
#Importing the libraries

import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as mtp
```

```python
#Importing the dataset

dataset=pd.read_csv("Datasets/Data_DataProcessing.csv")
print(dataset)
```

```python
#Separate the dependent(output) and independent variables(input)
x=dataset.iloc[:,:-1].values
```

```python
#input variable
x
```

```python
#output variable
y=dataset.iloc[:,3]
y
```

```python
from sklearn.impute import SimpleImputer
import numpy as np
```

```python
imputer=SimpleImputer(missing_values=np.nan,strategy='mean')
```

```python
#Fitting imputer object to the independent variable
imputer1=imputer.fit(x[:,1:3])
```

```python
#Replacing missing data with the calculated mean value
x[:,1:3]=imputer1.transform(x[:,1:3])
```

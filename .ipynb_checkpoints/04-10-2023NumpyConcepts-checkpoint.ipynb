```python
#To creating an ndarray
import numpy as np
arr=np.array([10,20,30,40])
print(arr)
print(type(arr))
print(np.__version__)
```

```python
#Dimensions in Array
```

```python
arr=np.array(10)
print(arr)
print(arr.ndim)
arr1=np.array([10,20,30,40])
print(arr1)
print(arr1.ndim)
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2)
print(arr2.ndim)
arr3=np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]])
print(arr3)
print(arr3.ndim)
```

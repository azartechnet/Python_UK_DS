```python
# Numerical stopwords refer to numbers or numerical expressions that do not add significant value to the understanding of the text and can be removed during text preprocessing. These are especially common in datasets involving reviews, comments, or any textual data where numbers might be prevalent but not necessarily informative for the task at hand.
```

```python
# Removing Numerical Stopwords
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download NLTK stopwords
nltk.download('stopwords')
nltk.download('punkt')

# Define sample text
text = """
I have 2 apples and 5 oranges. In 2023, the price of apples is expected to rise by 10%. 
The meeting is scheduled for 10:30 AM on the 15th of June.
"""
```

```python
#  Tokenization
# Tokenize the text into individual words.
```

```python
# Tokenize the text
words = word_tokenize(text)
print("Original Words:", words)
```

```python
# Removing Numerical Stopwords
# Use a regular expression to filter out numerical tokens.
```

```python
# Define a function to remove numerical stopwords
def remove_numerical_stopwords(words):
    pattern = re.compile(r'\b\d+\b|\b\d+\.\d+\b')
    filtered_words = [word for word in words if not pattern.match(word)]
    return filtered_words

# Remove numerical stopwords
filtered_words = remove_numerical_stopwords(words)
print("Filtered Words:", filtered_words)
```

```python
# Explanation and Rationale
# Library Installation: The nltk library is commonly used for various NLP tasks, including tokenization and stopword removal.

# Importing Libraries and Defining Sample Text: This step sets up the necessary imports and the sample text containing numerical expressions.

# Tokenization: The word_tokenize function from nltk splits the text into individual words. This is a common preprocessing step in NLP.

# Removing Numerical Stopwords:

# Regular Expression: The regular expression \b\d+\b|\b\d+\.\d+\b matches both integers and floating-point numbers. The \b word boundary ensures that only standalone numbers are matched.
# Filtering: A list comprehension filters out the numerical tokens from the list of words.
```

```python
# Single-character stopwords are typically individual letters or characters that do not add meaningful information to the text and can be removed during preprocessing to clean the data. Here are some examples and how to remove them in Python
```

```python
# Examples of Single-Character Stopwords
```

```python
# Common single-character stopwords include:

# Letters: a, b, c, etc.
# Other characters depending on the context: I (when used as a pronoun, it might not be removed), punctuation marks, etc.
```

```python
import nltk
from nltk.tokenize import word_tokenize

# Download NLTK punkt for tokenization
nltk.download('punkt')

# Define sample text
text = """
I have a cat. It is a black cat. The cat's name is B. I also have a dog named D. The letters a, b, and c are common in the English alphabet.
"""
```

```python
# Tokenization
# Tokenize the text into individual words.
```

```python
# Tokenize the text
words = word_tokenize(text)
print("Original Words:", words)
```

```python
# Removing Single-Character Stopwords
# Use a list comprehension to filter out single-character tokens
```

```python
# Define a function to remove single-character stopwords
def remove_single_character_stopwords(words):
    filtered_words = [word for word in words if len(word) > 1 or word.lower() == 'i']
    return filtered_words

# Remove single-character stopwords
filtered_words = remove_single_character_stopwords(words)
print("Filtered Words:", filtered_words)
```

```python
# Explanation and Rationale
# Library Installation: The nltk library is commonly used for various NLP tasks, including tokenization.

# Importing Libraries and Defining Sample Text: This step sets up the necessary imports and defines a sample text that contains single-character words.

# Tokenization: The word_tokenize function from nltk splits the text into individual words.

# Removing Single-Character Stopwords:

# List Comprehension: A simple list comprehension is used to filter out words that have a length of 1. Here, we keep the single-character word 'I' as it is a meaningful pronoun in English.
```

```python
#Contextual Stopwords
```

```python
# Contextual stopwords are words that are typically considered unimportant or irrelevant within a specific context or domain, even though they might not be traditional stopwords (like 'the', 'is', 'at'). These words can vary depending on the specific application, dataset, or problem being solved.
```

```python
# Examples of Contextual Stopwords
# In different contexts, the stopwords can change. Here are some examples for various contexts:

# Medical Texts:

# Common but uninformative terms: 'patient', 'doctor', 'medical', 'clinic'
# Example: "The patient was seen by the doctor at the clinic."
# Financial Reports:

# Common but uninformative terms: 'company', 'financial', 'report', 'quarter'
# Example: "The company's financial report for the quarter was released."
# News Articles:

# Common but uninformative terms: 'news', 'reporter', 'said', 'reported'
# Example: "The reporter said the news was reported yesterday."
# Social Media Posts:

# Common but uninformative terms: 'like', 'post', 'share', 'follow'
# Example: "Like and share this post to follow the updates."
```

```python
import nltk
from nltk.tokenize import word_tokenize

# Download NLTK punkt for tokenization
nltk.download('punkt')

# Define sample text
text = """
The company's financial report for the quarter was released yesterday. 
The patient was seen by the doctor at the clinic for a routine checkup.
Like and share this post to follow the updates.
"""
```

```python
# Tokenization
# Tokenize the text into individual words.
```

```python
# Tokenize the text
words = word_tokenize(text)
print("Original Words:", words)
```

```python
# Define Contextual Stopwords and Remove Them
# Create a list of contextual stopwords and filter them out from the tokenized words.
```

```python
# Define contextual stopwords for this example
contextual_stopwords = ['company', 'financial', 'report', 'quarter', 'patient', 'doctor', 'clinic', 'like', 'share', 'post', 'follow']

# Remove contextual stopwords
filtered_words = [word for word in words if word.lower() not in contextual_stopwords]
print("Filtered Words:", filtered_words)
```

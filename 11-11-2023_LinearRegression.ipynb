```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#1.DataCollections
```

```python
data=pd.read_csv('Datasets/BostonHousing.csv')
#data.head(10)
data.tail(10)
```

```python
#2.DataWrangling
#Input or independent variable
X=data.iloc[:,0:13]
X
```

```python
#output variable
Y=data['medv']
Y
```

```python
#3.DataAnalysis
```

```python
#Correlation
```

```python
import seaborn as sns
```

```python
#Creating a correlation matrix
correlation=data.corr()
#print(correlation)
```

```python
sns.heatmap(correlation)
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.show()
```

```python
#4.Training the algorithm
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=10)
```

```python
print(X_train.shape)
```

```python
X_train
```

```python
Y_train
```

```python
print(X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
#Creating the Model
lin_model=LinearRegression()
```

```python
#fit the data in to the model
lin_model.fit(X_train,Y_train)
#print(X_train)
#print(Y_train)
```

```python
#5.Test the algorithm
predicated_house_price=lin_model.predict(X_test)
```

```python
pd.DataFrame({'PredicatedValue':predicated_house_price,"KnownO/P":Y_test})
```
